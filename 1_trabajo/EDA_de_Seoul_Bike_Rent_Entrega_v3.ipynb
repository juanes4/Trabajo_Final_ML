**1.1 Descripción del caso:**

Seúl tiene un sistema de bicicletas compartidas llamado Ttareungyi, que es operado por la administración del gobierno local. Tienen estaciones repartidas por toda la ciudad para que las personas renten y devuelvan las bicicletas por hora. El problema es que la demanda puede variar mucho según la hora, el día de la semana, las condiciones climáticas y la estación. En verano en las horas pico faltan bicicletas y en invierno sobran. Al predecir la demanda podemos prever cuántas bicicletas debemos tener disponibles en la ciudad. Los datos son registros de cada hora y van desde diciembre de 2017 hasta noviembre de 2018. Son 8.760 filas en total, pero las horas en las que el sistema estuvo funcionando son 8.465.

**1.2 Pregunta de negocio:**

#### Pregunta de Negocio

¿Cómo podemos predecir la demanda de bicicletas por hora, con una proyección a futuro de 24 horas, considerando el clima, festivo, día de la semana y horas pico, para reducir escasez y oferta desmedida en el sistema?

**1.3 Reglas de negocio:**

-No se pueden poner más bicicletas de las que existen en la flota. Como no nos entregan el tamaño de la flota, asumimos que no hay restricción por ese lado y el modelo entrega la disponibilidad mínima que necesita el sistema.

-Solo se predice para días en que el sistema funciona.

-La redistribución tiene costo (camiones, personal) y capacidad limitada por turno, por eso se busca no tener que hacer reabastecimiento durante el día.

-La proyección es a 24 horas, así que el modelo depende del pronóstico del clima del día siguiente, que tiene su propio error (aunque a 24 horas es más preciso que a varios días).

-El servicio debe cubrir los picos laborales, porque perder esas rentas afecta a quien depende de la bicicleta para ir a trabajar.


**1.4 Funciones de los modelos (predictivo y prescriptivo):**

Usaremos DBSCAN como nuestro modelo descriptivo, nos va a ayudar a agrupar las horas en clusters, separando los datos en grupos que sean vecinos, y a identificar las horas atípicas (ruido).

MLP será nuestro modelo predictivo. Nos dirá cuántas bicicletas se rentarán en cada hora, teniendo en cuenta la hora, el clima, la estación, si es festivo y el día de la semana.

Y el modelo prescriptivo nos dirá cuántas bicicletas debemos tener disponibles cada día en toda la ciudad, usando el pico máximo de demanda predicho para ese día más un margen de seguridad.


**1.5 Criterio de negocio de evaluación:**

Reducir la cantidad de bicicletas sobrantes y faltantes. Actualmente sin el modelo podemos estimar cuántas bicicletas se van a rentar en cada hora basándonos en el promedio histórico de esa hora, y queremos reducir el faltante y el sobrante usando el modelo predictivo. La meta es que el total de bicicletas faltantes + sobrantes se reduzca en más de 20 % frente a planificar con el promedio de cada hora (estimado conservador). Se mide en el conjunto de prueba.

#### Imputación

Humedad nunca debería de ser 0.

Nuestro razonamiento para interpolar es que la humedad sigue un ciclo de 24 horas, por lo que no se puede interpolar hacia adelante simplemente, lo que haremos es basarnos en la hora del día anterior más cercano sin dato erróneo, y interpolar hacia adelante hasta la hora del día siguiente que no tenga dato erróneo.




```python
data_train["Humidity(%)"] = data_train["Humidity(%)"].replace(0,np.nan)#Remplazamos los valores 0 por nan para realizar interpolacion

data_train["Humidity(%)"] = data_train.groupby("Hour")["Humidity(%)"].transform(lambda x: x.interpolate(method='linear'))#Interpolamos en base a la hora, de forma lineal.
data_train['Humidity(%)'][data_train['Humidity(%)'] == 0]
```

#### 3.3 Distribución de cada variable numérica (histograma y diagrama de cajas) e identificación de valores atípicos

#### Outliers

Se identificaron outliers estadísticos (método IQR) en Rented Bike Count, Wind speed, Rainfall, Snowfall y Solar Radiation. Sin embargo, estos outliers no son errores de captura ni valores imposibles. Son eventos reales y esperables en este contexto (horas pico de demanda, eventos climáticos puntuales como tormentas o nevadas, y ráfagas de viento fuerte). Por lo tanto, no se recomienda eliminarlos ni tratarlos como datos erróneos, ya que hacerlo eliminaría información legítima y relevante para el análisis de demanda.

```python
# Diagramas de cajas de cada variable numérica para identificar valores atípicos
num_cols = data_train.select_dtypes(include=['float64', 'int64']).columns
unidades = {
    "Rented Bike Count": "bicicletas/hora", "Hour": "hora (0-23)", "Temperature(°C)": "°C",
    "Humidity(%)": "%", "Wind speed (m/s)": "m/s", "Visibility (10m)": "decámetros (10 m)",
    "Dew point temperature(°C)": "°C", "Solar Radiation (MJ/m2)": "MJ/m²",
    "Rainfall(mm)": "mm", "Snowfall (cm)": "cm"
}

n_cols = len(num_cols)
fig, axes = plt.subplots(nrows=(n_cols + 2) // 3, ncols=3, figsize=(15, 4 * ((n_cols + 2) // 3)))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    sns.boxplot(y=data_train[col], ax=axes[i], color='skyblue')
    axes[i].set_title(f'Distribución y Outliers: {col}', fontsize=10)
    axes[i].set_ylabel(unidades.get(col, col))

for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

**Conclusión outliers:**

La **variable Rented Bike Count** presenta una distribución inclinada hacia la derecha con la presencia de valores atípicos superiores a 2300 bicicletas rentadas por hora, los puntos aislados que se encuentran fuera de la caja que llegan hasta 3500 o más corresponde a horas con mayor demanda considerada fuera de lo normal posiblemente influenciada por otras variables.

La **variable Wind speed m/s** presenta un sesgo a la derecha, indicando valores atípicos, los puntos aislados se encuentran a partir de 4.2 m/s hasta alcanzar una velocidad de 7 m/s o más que corresponden a eventos meteorológicos reales como vientos muy fuertes aunque poco comunes.

#### 3.4 Frecuencias de las variables categóricas (gráfico de barras)

#### C.7 Clustering con DBSCAN

```python
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
import numpy as np
from sklearn.metrics import adjusted_rand_score

ks = np.arange(0.5, 6.4, 0.1)
inercia = []
silueta = []
unique_list = []
rand_index = []

RANDOM_STATE = 42
for e in ks:
    dbscan = DBSCAN(
    eps=e,
    min_samples=2,
    metric="euclidean"
    )


    labels = dbscan.fit_predict(data_model_scaled[variables_entrada])
    unique_list.append(np.unique(labels, return_counts=False))
    silueta.append(silhouette_score(data_model_scaled[variables_entrada], labels))
    rand_index.append(adjusted_rand_score(labels, labels))

resultados_kmeans = pd.DataFrame({
    "K": list(ks),
    "Silueta": silueta,
    "Rand_Index": rand_index
})
```

```python
count_list = [len(i) for i in unique_list]
```

**Conclusión:**

Con epsilon pequeño (menos de 1) DBSCAN encuentra muchísimos clusters y la silueta es baja. Entre 1.0 y 3.2 la silueta se estabiliza y el número de clusters va bajando. En 3.3 hay un salto de la silueta pero el número de clusters vuelve a bajar, porque a esa distancia DBSCAN empieza a unir muchos más en un mismo grupo. Nuestra elección fue epsilon = 3.2, justo antes de ese salto: la silueta está muy cerca a la que más tiene.

#### C.8 Gráfico de silueta

```python
from sklearn.metrics import silhouette_score, silhouette_samples

# Redefinimos localmente las variables con las que se ajustó el clustering
# original para evitar usar la lista global 'variables_entrada' que ahora incluye dummies de DBSCAN
variables_clustering = numerical_cols + cat_names
X_escalado = data_model_scaled[variables_clustering].values

# Silueta solo sobre los puntos que pertenecen a algún cluster: el ruido (-1)
# no es un grupo real y bajaría artificialmente el coeficiente
mascara = (data_model["Cluster_DBSCAN"] != -1).values
etiquetas = data_model.loc[mascara, "Cluster_DBSCAN"].values
X_sil = X_escalado[mascara]

if len(set(etiquetas)) >= 2:
    # Silueta media (resumen global) y silueta por punto (para ver cada cluster)
    sil_global = silhouette_score(X_sil, etiquetas)
    sil_puntos = silhouette_samples(X_sil, etiquetas)

    fig, ax = plt.subplots(figsize=(8, 5))
    y_inferior = 10
    colores = sns.color_palette("tab10", n_colors=len(set(etiquetas)))
    for i, c in enumerate(sorted(set(etiquetas))):
        # Se ordenan los valores de cada cluster para dibujar su "cuchillo" de silueta
        valores = np.sort(sil_puntos[etiquetas == c])
        y_superior = y_inferior + len(valores)
        ax.fill_betweenx(np.arange(y_inferior, y_superior), 0, valores, color=colores[i], alpha=0.8)
        ax.text(-0.05, y_inferior + 0.5 * len(valores), f"Cluster {c}")
        y_inferior = y_superior + 10  # espacio entre clusters para distinguirlos
    # La línea roja permite comparar cada cluster contra el promedio general
    ax.axvline(sil_global, color="red", linestyle="--", label=f"Silueta media = {sil_global:.3f}")
    ax.set_title("Gráfico de silueta por cluster DBSCAN (sin ruido)")
    ax.set_xlabel("Coeficiente de silueta (adimensional, de -1 a 1)")
    ax.set_ylabel("Observaciones agrupadas por cluster")
    ax.set_yticks([])
    ax.legend()
    plt.tight_layout()
    plt.show()

    print("Observaciones excluidas por ser ruido:", (~mascara).sum())
else:
    print("Se necesitan al menos 2 clusters (sin contar el ruido) para calcular la silueta. Ajuste eps y min_samples.")
```

**Conclusión:**

La silueta media es de 0.476, no está mal, y prácticamente ninguno tiene silueta negativa, o sea que casi no hay horas mal asignadas.

**4.2 Modelo predictivo (tipo, entradas, salida):**

**Tipo de modelo:** red neuronal MLP (MLPRegressor) dentro de un Pipeline, con StandardScaler para las entradas y la salida también estandarizada (TransformedTargetRegressor). El escalado se ajusta solo con los datos de entrenamiento.

**Variables de entrada:** Hour, Humidity, Temperature, Rainfall, Snowfall y las dummies de Seasons, Holiday y Day of week (18 variables).

**Variable de salida:** Rented Bike Count (bicicletas rentadas por hora).

**Por qué este modelo:** en el EDA vimos que la relación con la demanda no es lineal (la humedad tiene un rango ideal, la hora tiene dos picos y la forma del día cambia entre semana y fin de semana). Una red neuronal puede aprender esas combinaciones mejor que una regresión lineal.

**Hiperparámetros:** con búsqueda en rejilla y validación cruzada de 5 particiones, la mejor configuración fue 2 capas ocultas de 32 y 16 neuronas, activación ReLU y alpha = 0.01.

#### C.11 Modelo predictivo de regresión: red neuronal (MLP) con búsqueda en rejilla

Definiendo variables de entrada y variables de salida:

```python
# Definimos variables de entrada incluyendo las numéricas, las cat_names filtradas y los días de la semana codificados
variables_entrada = numerical_cols + cat_names + day_of_week_encoded_cols
variable_salida = "Rented Bike Count"
```

```python
X = data_model[variables_entrada]
y = data_model[variable_salida]
```

```python
#Ordenar los 2 datasets por fecha y hora
data_model = data_model.sort_values(["Date", "Hour"]).reset_index(drop=True)
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import Pipeline
from sklearn.compose import TransformedTargetRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

SEMILLA = 42
# División en entrenamiento y prueba usando la asignación corregida (x = características, y = objetivo)
X_ent, X_pru, y_ent, y_pru = train_test_split(X, y, test_size=0.25, random_state=SEMILLA)
print("Entrenamiento:", X_ent.shape, "| Prueba:", X_pru.shape)
```

```python
# Pipeline: escalado de entradas + red neuronal; la salida también se estandariza para facilitar el entrenamiento
red = TransformedTargetRegressor(
    regressor=Pipeline([
        ("escalador", StandardScaler()),
        ("mlp", MLPRegressor(solver="adam", max_iter=2000, random_state=42)),
    ]),
    transformer=StandardScaler(),
)

# Rejilla de hiperparámetros: arquitectura (capas y neuronas), activación y regularización
rejilla = {
    "regressor__mlp__hidden_layer_sizes": [(8,), (32,), (64,), (16, 16), (32, 16), (32, 32), (32, 16, 8)],
    "regressor__mlp__activation": ["relu", "identity"],   # identity = activación lineal
    "regressor__mlp__alpha": [1e-4, 1e-2],
}

# Configuramos GridSearchCV con múltiples métricas para evaluar tanto el RMSE como el R2
metricas = {
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2"
}

busqueda = GridSearchCV(
    red,
    rejilla,
    cv=5,
    scoring=metricas,
    n_jobs=-1,
    refit="RMSE"
)
busqueda.fit(X_ent, y_ent)

print("Mejores hiperparámetros:", busqueda.best_params_)
print(f"RMSE medio en validación cruzada (mejor modelo): {-busqueda.best_score_:.3f}")
print(f"R2 medio en validación cruzada (mejor modelo): {busqueda.cv_results_['mean_test_R2'][busqueda.best_index_]:.3f}")
```

```python
# Tabla de resultados de la rejilla
res = pd.DataFrame(busqueda.cv_results_)
tabla = pd.DataFrame({
    "capas": res["param_regressor__mlp__hidden_layer_sizes"].astype(str),
    "n_capas_ocultas": res["param_regressor__mlp__hidden_layer_sizes"].apply(len),
    "activacion": res["param_regressor__mlp__activation"],
    "alpha": res["param_regressor__mlp__alpha"],
    "RMSE_validacion": -res["mean_test_RMSE"],
    "desv_RMSE": res["std_test_RMSE"],
}).sort_values("RMSE_validacion").reset_index(drop=True)
display(tabla.head(10).round(3))

# Mejor RMSE por arquitectura y activación (mínimo sobre alpha)
resumen = tabla.groupby(["n_capas_ocultas", "capas", "activacion"])["RMSE_validacion"].min().unstack()
resumen.index = resumen.index.get_level_values("capas")
resumen.plot(kind="bar", figsize=(9, 4))
plt.title("RMSE de validación por arquitectura y función de activación")
plt.xlabel("Neuronas por capa oculta")
plt.ylabel("(validación cruzada)")
plt.xticks(rotation=20)
plt.legend(title="Activación")
plt.tight_layout()
plt.show()
```

**Conclusión:**

La activación lineal (identity) da un RMSE de casi 440 en todas las arquitecturas, porque una red lineal funciona igual que una regresión lineal y no puede captar los picos de demanda. Con ReLU el error baja mucho en todas las arquitecturas. Las mejores son las de 2 capas con 32 neuronas en la primera: (32, 16) con RMSE de 232.7 y (32, 32) con 237.9. Antes de que incluyéramos el día de la semana el mejor RMSE era 253.6, entonces el día de la semana mejoró el error.

**4.3 Modelo prescriptivo (tipo, entradas, salida):**

**Tipo:** reglas de decisión sobre las predicciones del modelo predictivo.

**Regla:** Disponibilidad del día = pico horario máximo predicho para ese día + margen de seguridad teniendo en cuenta el error del modelo.

**Entradas:** la demanda horaria predicha por el MLP para las próximas 24 horas (con el clima pronosticado, la hora, la estación, si es festivo y el día de la semana) y el margen de seguridad calculado con los errores del modelo en el pico diario.

**Salida:** la cantidad de bicicletas que debe estar disponible cada día y la hora en que se espera el pico.

Asumimos que como no nos entregan el tamaño de la flota, no hay restricciones por ese lado y el modelo nos entregará la disponibilidad mínima requerida por día. Así se cumple la regla de negocio de no tener que hacer reabastecimiento durante el día.


#### 4.4 Implementación del modelo prescriptivo

**Variables de entrada:** la demanda horaria predicha por la red MLP para las 24 horas de cada día (columna `Demanda_predicha`) y el margen de seguridad calculado con el error del modelo en el pico diario.

**Variable de salida:** `Disponibilidad_recomendada`, las bicicletas que deben estar disponibles en la ciudad ese día, y `Hora_pico_predicha`, la hora en que esperamos el pico.

Para armar días completos necesitábamos una predicción de cada hora del año hecha por un modelo que no hubiera visto esa hora. Por eso usamos `cross_val_predict` con la mejor configuración de la búsqueda en rejilla. Las predicciones negativas las llevamos a 0, porque no existen rentas negativas.

```python
# Predice cada hora del año con un modelo que no la vio (validación cruzada de 5 particiones) usando la mejor configuración de la red
from sklearn.model_selection import cross_val_predict, KFold
from sklearn.base import clone

pred_oof = cross_val_predict(
    clone(busqueda.best_estimator_),
    data_model[variables_entrada],
    data_model[variable_salida],
    cv=KFold(n_splits=5, shuffle=True, random_state=SEMILLA),
)
data_model["Demanda_predicha"] = np.clip(pred_oof, 0, None)
print(f"RMSE de las predicciones fuera de muestra: {np.sqrt(mean_squared_error(data_model[variable_salida], data_model['Demanda_predicha'])):.1f} bicicletas/hora")
```

```python
# Resume cada día: pico real, pico predicho y la hora en que ocurre cada uno
idx_real = data_model.groupby("Date")[variable_salida].idxmax()
idx_pred = data_model.groupby("Date")["Demanda_predicha"].idxmax()

dias = pd.DataFrame({
    "Seasons": data_model.loc[idx_real, "Seasons"].values,
    "Pico_real": data_model.loc[idx_real, variable_salida].values,
    "Hora_pico_real": data_model.loc[idx_real, "Hour"].values,
    "Pico_predicho": data_model.loc[idx_pred, "Demanda_predicha"].values,
    "Hora_pico_predicha": data_model.loc[idx_pred, "Hour"].values,
}, index=idx_real.index)

print("Días con servicio:", len(dias))
print(f"Error absoluto medio en el pico diario: {(dias['Pico_real'] - dias['Pico_predicho']).abs().mean():.1f} bicicletas")
print(f"Días en que la hora del pico se acierta con ±1 hora: {100 * ((dias['Hora_pico_real'] - dias['Hora_pico_predicha']).abs() <= 1).mean():.1f} %")
display(dias.head())
```

**Regla de decisión:**

`Disponibilidad_recomendada = Pico_predicho del día + margen`

Tomamos como margen el percentil 95 del error en el pico diario (pico real − pico predicho), calculado con el 75 % de los días. Así la disponibilidad debería cubrir el pico en el 95 % de los días sin reabastecer durante el día. El 25 % de días restante lo usamos para evaluar la regla.

```python
# Calcula el margen de seguridad con el 75 % de los días y aplica la regla de disponibilidad al 25 % restante
NIVEL_SERVICIO = 0.95
dias_ent, dias_pru = train_test_split(dias, test_size=0.25, random_state=SEMILLA)
dias_pru = dias_pru.copy()

margen = np.quantile(dias_ent["Pico_real"] - dias_ent["Pico_predicho"], NIVEL_SERVICIO)
dias_pru["Disponibilidad_recomendada"] = np.ceil(dias_pru["Pico_predicho"] + margen).astype(int)

print(f"Margen de seguridad (percentil {int(NIVEL_SERVICIO * 100)} del error en el pico): {margen:.0f} bicicletas")
display(dias_pru.sort_index().head(10))
```

```python
# Función prescriptiva: recibe las 24 predicciones horarias de un día y devuelve la disponibilidad recomendada y la hora del pico
def recomendar_disponibilidad(predicciones_24h, margen=margen):
    predicciones_24h = np.clip(np.asarray(predicciones_24h), 0, None)
    hora_pico = int(np.argmax(predicciones_24h))
    disponibilidad = int(np.ceil(predicciones_24h.max() + margen))
    return {"disponibilidad_recomendada": disponibilidad, "hora_pico": hora_pico,
            "pico_predicho": round(float(predicciones_24h.max()))}

# Ejemplo: un miércoles de verano, 22 °C, sin lluvia (en producción el clima vendría del pronóstico del día siguiente)
dia_ejemplo = pd.DataFrame({
    "Hour": range(24), "Humidity(%)": 55.0, "Temperature(°C)": 22.0, "Rainfall(mm)": 0.0, "Snowfall (cm)": 0.0,
})
for col in cat_names + day_of_week_encoded_cols:
    dia_ejemplo[col] = 0.0
dia_ejemplo[["Seasons_Summer", "Holiday_No Holiday", "Day of week_Wednesday"]] = 1.0

print(recomendar_disponibilidad(busqueda.best_estimator_.predict(dia_ejemplo[variables_entrada])))
```

**Criterio de negocio (1.5): bicicletas faltantes y sobrantes con y sin el modelo**

```python
# 1. Estrategias de Planificación (Oferta de Bicicletas)
promedio_hora = y_ent.groupby(X_ent["Hour"]).mean()
plan_sin_modelo = X_pru["Hour"].map(promedio_hora).values
plan_con_modelo = y_pred

# 2. Función de Evaluación de Demanda No Cubierta y Excedente
def evaluar_plan(real, plan):
    diferencia = np.asarray(real) - np.asarray(plan)
    faltantes = diferencia.clip(min=0).sum()
    sobrantes = (-diferencia).clip(min=0).sum()
    return faltantes, sobrantes

# 3. Cálculo de Métricas por Estrategia
f_sin, s_sin = evaluar_plan(y_pru, plan_sin_modelo)
f_con, s_con = evaluar_plan(y_pru, plan_con_modelo)

# 4. Tabla Comparativa de Resultados de Negocio
criterio = pd.DataFrame({
    "Sin Modelo (Promedios)": [f_sin, s_sin, f_sin + s_sin],
    "Con Modelo (Red MLP)": [f_con, s_con, f_con + s_con]
}, index=["Bicicletas faltantes", "Bicicletas sobrantes", "Total Desviación"])

criterio["Reducción (%)"] = 100 * (1 - criterio["Con Modelo (Red MLP)"] / criterio["Sin Modelo (Promedios)"])
display(criterio.round(1))

# 5. Gráfico de Barras Comparativo
plt.figure(figsize=(8, 4.5))
criterio.iloc[:2, :2].T.plot(kind="bar", ax=plt.gca(), color=["#e74c3c", "#3498db"], edgecolor="black", alpha=0.8)
plt.title("Bicicletas Faltantes vs Sobrantes (Conjunto de Prueba)", fontsize=11, fontweight="bold")
plt.ylabel("Cantidad de Bicicletas")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.legend(["Faltantes", "Sobrantes"])
plt.tight_layout()
plt.show()
```

Con el modelo el faltante baja a 156,863 y el sobrante a 155,939, en total 312,802, o sea 64.3% menos, mucho más que la métrica de 20% que pusimos en el criterio de negocio. #### 5.3 Evaluación del modelo prescriptivo

Comparamos la regla con el modelo contra planear sin modelo, tomando como pico del día el mayor promedio histórico por hora (calculado con los datos de entrenamiento) más su propio margen del percentil 95. Las dos estrategias buscan cubrir el pico el 95 % de los días, y gana la que lo logra con menos bicicletas sobrantes.

```python
# Compara la disponibilidad diaria con y sin modelo: cobertura del pico, bicicletas sobrantes y faltantes
pico_sin_modelo = promedio_hora.max()
margen_sin = np.quantile(dias_ent["Pico_real"] - pico_sin_modelo, NIVEL_SERVICIO)
dias_pru["Disponibilidad_sin_modelo"] = np.ceil(pico_sin_modelo + margen_sin).astype(int)

def evaluar_disponibilidad(real, disponible):
    real, disponible = np.asarray(real), np.asarray(disponible)
    return {
        "Días con el pico cubierto (%)": 100 * (real <= disponible).mean(),
        "Bicicletas sobrantes en el pico (promedio por día)": np.clip(disponible - real, 0, None).mean(),
        "Bicicletas faltantes en el pico (total)": np.clip(real - disponible, 0, None).sum(),
    }

comparacion_presc = pd.DataFrame({
    "Sin modelo (promedio histórico)": evaluar_disponibilidad(dias_pru["Pico_real"], dias_pru["Disponibilidad_sin_modelo"]),
    "Con modelo (MLP + margen)": evaluar_disponibilidad(dias_pru["Pico_real"], dias_pru["Disponibilidad_recomendada"]),
})
comparacion_presc["Cambio (%)"] = 100 * (comparacion_presc.iloc[:, 1] / comparacion_presc.iloc[:, 0] - 1)
print(f"Disponibilidad fija sin modelo: {dias_pru['Disponibilidad_sin_modelo'].iloc[0]} bicicletas todos los días | días evaluados: {len(dias_pru)}")
display(comparacion_presc.round(1))

# Pico real de cada día de prueba frente a la disponibilidad recomendada con y sin modelo
orden = dias_pru.sort_index()
plt.figure(figsize=(14, 5))
plt.plot(orden.index, orden["Pico_real"], "o", ms=4, color="black", label="Pico real del día")
plt.plot(orden.index, orden["Disponibilidad_recomendada"], color="#3498db", label="Disponibilidad con modelo")
plt.plot(orden.index, orden["Disponibilidad_sin_modelo"], "--", color="#e74c3c", label="Disponibilidad sin modelo")
plt.title("Disponibilidad diaria recomendada vs pico real (días de prueba)")
plt.xlabel("Fecha")
plt.ylabel("Bicicletas (unidades)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Conclusión:**

Con un margen de 633 bicicletas, el modelo cubrió el pico en el 96,6 % de los 89 días de prueba. En promedio sobraron 514 bicicletas por día. Sin modelo, para cubrir el 95,5 % de los días habría que tener 3.172 bicicletas diarias, con 1.609 sobrantes. Con el modelo las bicicletas ociosas bajaron 68 % y las que faltaron en el pico bajaron 77 % (de 749 a 174). Además, el modelo acertó la hora del pico con ±1 hora en el 74 % de los días, y eso sirve para programar la redistribución. La simulación usó el clima real y no el pronóstico, y calculamos la disponibilidad para toda la ciudad y no por estación.

DBSCAN encontró 8 grupos de horas según la estación y el festivo (silueta 0,476) y 14 horas de ruido con clima extremo. La demanda más baja quedó en invierno (≈ 200 bicicletas/hora) y la más alta en verano y otoño (más de 900).

La MLP ganadora tiene 2 capas ocultas (32, 16), ReLU y alpha 0,01. En prueba logró un RMSE de 227,4, un MAE de 147,8 bicicletas/hora y un R² de 0,869. Su error es casi tres veces menor que el de la línea base (RMSE 628,6). En validación cruzada el RMSE fue 232,7, así que no vemos sobreajuste.

Frente a planear con el promedio de cada hora, las bicicletas faltantes más sobrantes bajaron 64,3 %, por encima de la meta del 20 % del criterio 1.5. La regla de disponibilidad (pico predicho + 633 bicicletas) cubrió el pico en el 96,6 % de los días y redujo 68 % las bicicletas ociosas.

**Limitaciones y mejoras.** El modelo se queda corto en los picos más altos. La partición fue aleatoria por horas y no por fecha. Usamos el clima real en vez del pronóstico, y los datos son de un solo año. Como mejoras proponemos validar con una partición temporal, usar el pronóstico del clima, agregar la demanda de horas anteriores y predecir por estación o zona.
